# Method

Method 是重点部分。

你的 Method 可以包含：

## 1. Dataset and Label Construction

说明：

- 使用 Mouse Grimace Faces 数据
- 根据 MGS 分数构建标签
- `mgs_mean >= 1` 作为 impaired
- `mgs_mean < 1` 作为 well-being
- 去掉无效分数，例如 `-` 或 `9`

## 2. Group-Aware Train/Validation Split

说明：

你使用 `GroupShuffleSplit`，并且按照 mouse identity 或 subset/id 分组。

目的：

> prevent identity leakage between training and validation sets.

也就是说，同一只老鼠的图片不能同时出现在 train 和 validation 中。

## 3. Image Preprocessing

说明：

- Resize to 224×224
- Normalize with ImageNet statistics
- Data augmentation
- Muzzle crop 或 manual crop

## 4. Model and Training

说明：

- 使用的模型，例如 EfficientNet-B0
- Loss function，例如 CrossEntropyLoss
- Optimizer，例如 Adam
- Learning rate
- Batch size
- Epochs
- Class weights
- Early stopping
- ReduceLROnPlateau
- Best model checkpoint

# Experiments

Experiments 部分回答：

> 你如何验证自己的方法？

可以设计为：

## Experiment 1: Full Image Baseline

目的：

测试完整图像能不能分类。

## Experiment 2: Muzzle-Only Classification

目的：

测试 muzzle region 是否包含足够信息。

## Experiment 3: No-Muzzle Ablation

目的：

把 muzzle 区域模糊或遮挡，看模型性能是否下降。

如果性能明显下降，说明 muzzle 对分类很重要。

## Experiment 4: Saliency / Grad-CAM Analysis

目的：

检查模型到底关注图片哪里。

# Results

Results 部分应该放核心结果图和表格。

建议包括：

## 1. Label Distribution

说明数据是否平衡。

## 2. Training Curve

展示 train loss、validation metric 是否稳定。

## 3. Confusion Matrix

说明模型主要错在哪里。

## 4. Classification Report

包括：

- Precision
- Recall
- F1-score
- Macro F1
- Weighted F1

## 5. ROC Curve and AUC

说明模型区分 well-being 和 impaired 的能力。

## 6. Ablation Comparison Table

例如：

```text
Full image
Muzzle only
No muzzle / blurred muzzle
```

比较它们的 Accuracy、Macro F1、AUC。

---

# Discussion

Discussion 不要只是重复结果。

这里要解释：

- 为什么 muzzle-only 有效？
- 为什么 full image baseline 可能受到背景干扰？
- 为什么 no-muzzle ablation 性能下降？
- Grad-CAM 是否支持你的假设？
- 你的实验有什么限制？

例如可以讨论：

> The muzzle-only model performed better than the full-image baseline, suggesting that reducing irrelevant background information helps the model focus on pain-related facial features. The no-muzzle ablation further supports this interpretation because performance decreased when the muzzle region was removed or blurred.

也要写 limitations：

- Manual crop 可能不适合大规模应用
- Dataset size still limited
- Binary label loses fine-grained pain severity information
- Generalization to unseen settings is not fully verified

# Conclusion

Conclusion 要简洁。

可以回答你的主问题：

然后写 future work：

- Automatic muzzle detection
- Larger dataset
- More robust external validation
- Multi-class or regression prediction for MGS score
- Better localization methods
